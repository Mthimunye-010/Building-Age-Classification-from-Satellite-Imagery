```python
import numpy as np

bins = [-np.inf, 1945, 1984, 2004, np.inf] #Bins has 5 edges for 4 classes, -np.inf & np.inf are the edges so no year falls out the nins

df["age_class"] = pd.cut(df["weighted_mean_year"], bins=bins, labels=[1, 2, 3, 4]).astype(int) #puts each weighted mean year in bins and labels it 1-
#astype turns the labels into plain intergers                                                   #the rigth edge is included so that a building built exactly in 1984
                                                                                               #falls in class 2, not 3


pixels = df.drop_duplicates(["px_key", "py_key"]) #Keeps one row per pixel, so the counts are pixels rather than pixel-years
print(pixels["age_class"].value_counts().sort_index())
```

```python
qa = df[["qa_valid_1", "qa_valid_2", "qa_valid_3"]]

print(qa.mean()) #share of rows valid in each slot
print(qa.any(axis=1).mean())  #share with at least one valid slot
```

```python
df_mad = pd.read_parquet("madrid_train.parquet")
df_mad.shape
```

```python
bins_mad = [-np.inf, 1960, 1984, 2004, np.inf]
df_mad["age_class"] = pd.cut(df_mad["weighted_mean_year"], bins=bins_mad, labels=[1, 2, 3, 4]).astype(int)

pix_mad = df_mad.drop_duplicates(["px_key", "py_key"])

print(pix_mad["age_class"].value_counts().sort_index())
```

```python
print(df_mad.shape, df_mad["city"].unique())

qa = df_mad[["qa_valid_1", "qa_valid_2", "qa_valid_3"]]

print(qa.mean())
print(qa.any(axis=1).mean())

for n in [1, 2, 3]:
    d = df_mad.loc[df_mad[f"qa_valid_{n}"], f"doy_{n}"]
    print(n, d.between(152, 244).mean().round(3), d.min(), d.max())
```

```python
def summer_share(d):
    ok = pd.DataFrame({n: d[f"qa_valid_{n}"] & d[f"doy_{n}"].between(152, 244) for n in [1, 2, 3]}) #ok marks, for each slot whether the reading is valid & from June-Aug
    return ok.any(axis=1).mean() #any(axis=1) asks if atleast 1 of the 3 slots qualifies
                         #mean() gives the share of rows where its true
print("Amsterdam:", round(summer_share(df), 3))
print("Madrid:  ", round(summer_share(df_mad), 3))
```

```python
#NoteBook 3 checkings:
#1.Which reading does it actually use? our season finding says "first valid reading" can come from any month.
#This check shows what the pipeline really sees.

first = np.where(df["qa_valid_1"], 1, np.where(df["qa_valid_2"], 2, 3))
#"first" records which slot would be chosen for each row
doy = np.where(first == 1, df["doy_1"], np.where(first == 2, df["doy_2"], df["doy_3"]))
#"doy" is the day of the year of that chosen reading.

print(pd.Series(first).value_counts(normalize=True).sort_index().round(3))
#1st print shows how often each slot gets used
print("share in June-Aug:", pd.Series((doy >= 152) & (doy <=244)).mean().round(3))
#2nd print shows the share of chosen readings that come from summer.
```

```python
def first_valid_summary(d):
    first = np.where(d["qa_valid_1"], 1, np.where(d["qa_valid_2"], 2, 3))
    doy = np.where(first == 1, d["doy_1"], np.where(first == 2, d["doy_2"], d["doy_3"]))
    print(pd.Series(first).value_counts(normalize=True).sort_index().round(3).to_dict())
    print("share in Jun-Aug:", pd.Series((doy >= 152) & (doy <= 244)).mean().round(3))

cols = ["qa_valid_1", "qa_valid_2", "qa_valid_3", "doy_1", "doy_2", "doy_3"]
src = df_mad if "df_mad" in globals() else pd.read_parquet("madrid_train.parquet", columns=cols)
first_valid_summary(src)

    
```
